```python
import random as r

class Account:
    account_count = 0

    def __init__(self, depositor, balance):
        self.bank = "SC은행"
        self.account_num = f"{str(r.randint(0, 999)).zfill(3)}-{str(r.randint(0,99)).zfill(2)}-{str(r.randint(0,999999)).zfill(6)}"
        self.depositor = depositor
        self.balance = int(balance)
        self.deposit_history_list = []
        self.withdrawal_history_list = []
        self.deposit_count = 0

        Account.account_count += 1
        
    # Account 클래스로부터 생성된 계좌의 개수 출력
    def get_account_num(self):
        print(f"총 계좌 개수: {Account.account_count}개, 내 계좌: {self.account_num}" )

    # 입금 메서드
    def deposit(self, deposit_amount):
        if deposit_amount < 1:
            print("최소 입금 금액은 1원 이상만 가능합니다.")
            return
        
        if deposit_amount >= 1:
            self.balance += deposit_amount
            self.deposit_count += 1
            self.deposit_history_list.append(f"입금 {deposit_amount:,}원")
            print(f"{deposit_amount:,}원을 입금하였습니다. \n현재 잔액 {self.balance:,}원입니다.")
            print(f"현재 입금 횟수 : {self.deposit_count}회입니다.")

        if self.deposit_count % 5 == 0:
            interest = int(self.balance * 0.01)
            self.balance += interest
            self.deposit_history_list.append(f"이자 {deposit_amount:,}원")
            print(f"🎉 입금 5회 달성! 이자 {interest:,}원이 지급되었습니다.")
            print(f"이자 지급 후 잔액 {self.balance:,}원입니다.")

    # 출금 메서드
    def withdraw(self, withdrawal_amount):
        if withdrawal_amount < 1:
            print("최소 출금 금액은 1원 이상만 가능합니다.")
            return
        
        if withdrawal_amount > self.balance:
            print(f"출금 불가능합니다. \n현재 잔액 {self.balance:,}원입니다. ")
            return
        
        if withdrawal_amount <= self.balance:
            self.balance -= withdrawal_amount
            self.withdrawal_history_list.append(f"출금 {withdrawal_amount:,}원")
            print(f"{withdrawal_amount:,}원을 출금하였습니다. \n현재 잔액 {self.balance:,}원입니다.")

    # 입금 내역 기록
    def deposit_history(self):
        if len(self.deposit_history_list) == 0:
            print("입금 내역이 없습니다.")
            return
        else:
            print(f"입금내역: {self.deposit_history_list}")

    # 출금 내역 기록
    def withdrawal_history(self):
        if len(self.withdrawal_history_list) == 0:
            print("출금 내역이 없습니다.")
            return
        else:
            print(f"출금내역: {self.withdrawal_history_list}")

    # 정보 출력 메서드
    def display_info(self):
        print(f"은행이름: {self.bank}, 예금주: {self.depositor}, 계좌번호: {self.account_num}, 잔고 : {self.balance:,}")
```

```python
p1 = Account("파이썬", 20000)
p2 = Account("자네", 1200000)
p3 = Account("정말", 700000)
p4 = Account("이럴", 2500000)
p5 = Account("텐가", 1800000)

people = [p1, p2, p3, p4, p5]
```
